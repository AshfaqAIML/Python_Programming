## What Is Tkinter?

Tkinter is a Python library for creating GUI applications.

It is commonly available with standard Python installations.

A simple Tkinter program looks like:

```python id="ay0m5q"
import tkinter as tk

window = tk.Tk()

window.mainloop()
```

This creates a basic application window.

---

## Understanding the Code

The first line:

```python id="z3x9k6"
import tkinter as tk
```

imports Tkinter.

Then:

```python id="c5m2q8"

```

creates the main application window.

Finally:

```python id="r7p4n1"
window.mainloop()
```

starts the GUI event loop.

The program keeps running and waits for user actions such as:

* Clicking a button
* Typing text
* Closing the window

# 6. The `pack()` Geometry Manager

Tkinter needs to know **where widgets should appear**.

One simple geometry manager is:

```python id="q6m3x8"
pack()
```

Example:

```python id="j5x8p1"
label = tk.Label(root, text="Hello")
label.pack()
```

Without:

```python id="acq80i"
pack()
```

the widget may not appear as expected.

---

## Simple Mental Model

Think of `pack()` as saying:

> Put this widget inside the parent window using Tkinter's packing rules.

Example:

```python id="y4m2q9"
label1 = tk.Label(root, text="First")
label1.pack()

label2 = tk.Label(root, text="Second")
label2.pack()
```

The labels will normally appear one after another vertically.

# 13. Getting Text from `Entry`

### Radiobutton

and:

```python id="x4m7p2"
listbox.get(index)
```

Example:

```python id="m9q2x8"
import tkinter as tk

root = tk.Tk()

listbox = tk.Listbox(root)

for language in ["Python", "Java", "C++"]:
    listbox.insert(tk.END, language)

listbox.pack()

def show_selection():
    selection = listbox.curselection()

    if selection:
        value = listbox.get(selection[0])
        print("Selected:", value)

button = tk.Button(
    root,
    text="Select",
    command=show_selection
)

button.pack()

root.mainloop()
```

If Python is selected:


This demonstrates how a GUI can provide basic text-editing functionality.

the GUI will not behave as a running application.

---

## Mistake 2: Calling the Button Function Immediately

Incorrect:

```python id="x8m3p7"
button = tk.Button(
    root,
    text="Click",
    command=greet()
)
```

This calls `greet()` immediately.

Correct:

```python id="m4q9x2"
button = tk.Button(
    root,
    text="Click",
    command=greet
)
```

Tkinter will call `greet` when the button is clicked.

---

## Mistake 3: Forgetting to Place Widgets

Creating:

```python id="p7x2m8"
label = tk.Label(root, text="Hello")
```

does not automatically place it visibly.

Use a geometry manager such as:

```python id="q4m9x1"
label.pack()
```

---

## Mistake 4: Confusing `Entry` and `Text`

Use:

```text id="v6m3x9"
Entry
```

for one-line input.

Use:

```text id="p2q8m5"
Text
```

for multi-line text.

---

## Mistake 5: Using Checkbutton When Only One Choice Is Needed

Use:

```text id="k3x7m1"
Radiobutton
```

when the user should choose one option from a group.

Use:

```text id="m8q4p2"
Checkbutton
```

when multiple independent options can be selected.
