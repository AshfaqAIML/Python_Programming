# 1. Introduction to RegEx

A regular expression is a **pattern that describes text we want to find**.

Suppose:

```python
text = "Python is easy"
```

If we want to search for:

```text
Python
```

we can use:

```python
import re

result = re.search("Python", text)

print(result)
```

Output will contain a match object similar to:

```text
<re.Match object; span=(0, 6), match='Python'>
```

This means Python found the pattern.

# 4. The `search()` Function

The `search()` function searches the entire string and returns the **first matching occurrence**.

Example:

```python
import re

text = "I am learning Python."

result = re.search("Python", text)

print(result.group())
```

Output:

```text
Python
```

The `group()` method returns the actual matched text.

# 5. The Dot Metacharacter

The dot:

```text
.
```

is a special regex character called a **metacharacter**.

It normally matches **any single character**, except newline characters.

Example:

```python
import re

text = "cat"

result = re.search("c.t", text)

print(result.group())
```

Output:

```text
cat
```

The pattern:

```text
c.t
```

means:

```text
c
any one character
t
```

---

# 6. Dot Metacharacter Examples

The pattern:

```text
c.t
```

can match:

```text
cat
cot
cut
c9t
c-t
```

Example:

```python
import re

words = ["cat", "cot", "cut", "c9t"]

for word in words:
    if re.search("c.t", word):
        print(word)
```

Output:

```text
cat
cot
cut
c9t
```

---

## Dot Does Not Match Newline by Default

Consider:

```python
text = "Hello\nWorld"
```

The `.` normally does not match the newline character unless appropriate regex flags are used.

For beginners, remember:

> **`.` usually means any single character except a newline.**

# 21. Quantifiers

Quantifiers tell regex **how many times a pattern should occur**.

Important quantifiers include:

| Quantifier | Meaning             |
| ---------- | ------------------- |
| `*`        | Zero or more        |
| `+`        | One or more         |
| `?`        | Zero or one         |
| `{n}`      | Exactly `n`         |
| `{n,}`     | At least `n`        |
| `{n,m}`    | Between `n` and `m` |

---

# 22. `*` — Zero or More

The `*` quantifier means:

> Match the previous pattern zero or more times.

Example:

```python
import re

text = "ct"

result = re.search(r"ca*t", text)

print(result.group())
```

Output:

```text
ct
```

The `a` occurs zero times.

It would also match:

```text
cat
caat
caaat
```

---

# 23. `+` — One or More

The `+` quantifier means:

> Match the previous pattern one or more times.

Example:

```python
import re

text = "caaat"

result = re.search(r"ca+t", text)

print(result.group())
```

Output:

```text
caaat
```

It requires at least one `a`.

Therefore:

```text
cat
caat
caaat
```

can match.

But:

```text
ct
```

cannot match.

---

# 24. `?` — Zero or One

The `?` quantifier means:

> The previous pattern can occur zero or one time.

Example:

```python
import re

text = "color"

result = re.search(r"colou?r", text)

print(result.group())
```

Output:

```text
color
```

This pattern can match both:

```text
color
colour
```

because the `u` is optional.

---

# 38. `finditer()`

The `finditer()` function also finds all matches, but instead of returning a simple list, it returns an iterator containing **match objects**.

Example:

```python
import re

text = "Python is easy. Python is powerful."

matches = re.finditer(r"Python", text)

for match in matches:
    print(match.group())
```

Output:

```text
Python
Python
```

---

# 39. Getting Match Positions with `finditer()`

One advantage of `finditer()` is that we can access the position of each match.

```python
import re

text = "Python is easy. Python is powerful."

matches = re.finditer(r"Python", text)

for match in matches:
    print("Match:", match.group())
    print("Start:", match.start())
    print("End:", match.end())
```

Possible output:

```text
Match: Python
Start: 0
End: 6
Match: Python
Start: 16
End: 22
```

The exact positions depend on the input text.

---

# 40. `findall()` vs `finditer()`

| `findall()`                           | `finditer()`                             |
| ------------------------------------- | ---------------------------------------- |
| Returns a list                        | Returns an iterator                      |
| Simple results                        | Match objects                            |
| Easy when only matched text is needed | Useful when positions/details are needed |

Use:

```python
re.findall()
```

when you simply want all matching values.

Use:

```python
re.finditer()
```

when you need information such as:

```text
matched text
start position
end position
```

# 44. Replacing Words Case-Insensitively

By default, regex matching is usually case-sensitive.

For example:

```python
import re

text = "Python PYTHON python"

new_text = re.sub(r"python", "Java", text)

print(new_text)
```

Output:

```text
Python PYTHON Java
```

Only lowercase `"python"` matched.

We can use `re.IGNORECASE`:

```python
import re

text = "Python PYTHON python"

new_text = re.sub(
    r"python",
    "Java",
    text,
    flags=re.IGNORECASE
)

print(new_text)
```

Output:

```text
Java Java Java
```

# 45. The `compile()` Function

The `compile()` function creates a reusable regular expression pattern.

Example:

```python
import re

pattern = re.compile(r"\d+")

text = "I have 10 apples and 20 oranges."

matches = pattern.findall(text)

print(matches)
```

Output:

```text
['10', '20']
```

---

# 46. Why Use `compile()`?

If the same regex pattern is going to be used multiple times, compiling it can make the code cleaner.

Example:

```python
import re

pattern = re.compile(r"\d+")

text1 = "Age: 25"
text2 = "Score: 95"

print(pattern.findall(text1))
print(pattern.findall(text2))
```

Output:

```text
['25']
['95']
```

Instead of repeatedly writing:

```python
re.findall(r"\d+", text)
```

we reuse:

```python
pattern
```

# 48. Exercise: Finding Valid Emails

One common practical use of regular expressions is checking whether text looks like an email address.

Consider:

```text
kamraan@example.com
```

A simple email pattern could be:

```text
[\w.-]+@[\w.-]+\.\w+
```

This is a useful beginner-level pattern, but real-world email validation can be much more complicated.

---

# 49. Understanding the Email Pattern

Consider:

```text
[\w.-]+@[\w.-]+\.\w+
```

Let's break it down.

### `[\w.-]+`

Matches one or more characters such as:

```text
letters
digits
underscore
dot
hyphen
```

This represents the username portion.

For example:

```text
kamraan
kamraan123
kamraan.test
```

---

### `@`

Matches the literal `@` character.

---

### `[\w.-]+`

Matches the domain name.

Examples:

```text
gmail
example
my-company
```

---

### `\.`

Matches a literal dot.

We escape the dot because:

```text
.
```

has a special meaning in regex.

---

### `\w+`

Matches one or more word characters.

This can represent:

```text
com
org
net
in
```

---

# 50. Simple Email Matching

```python
import re

email = "kamraan@example.com"

pattern = r"[\w.-]+@[\w.-]+\.\w+"

if re.fullmatch(pattern, email):
    print("Valid email format")
else:
    print("Invalid email format")
```

Output:

```text
Valid email format
```

---

# 51. Invalid Email Examples

```python
import re

pattern = r"[\w.-]+@[\w.-]+\.\w+"

emails = [
    "kamraan@example.com",
    "user@gmail.com",
    "invalid-email",
    "hello@",
    "@example.com",
    "user@example"
]

for email in emails:
    if re.fullmatch(pattern, email):
        print(email, "→ Valid")
    else:
        print(email, "→ Invalid")
```

Output:

```text
kamraan@example.com → Valid
user@gmail.com → Valid
invalid-email → Invalid
hello@ → Invalid
@example.com → Invalid
user@example → Invalid
```

Again, this is a **simple format check**, not a complete RFC-level email validator.

---

# 52. Why Use `fullmatch()` for Validation?

Suppose we use:

```python
re.search(pattern, email)
```

It can find a matching portion somewhere inside the string.

For validation, we usually want:

> The entire input should follow the pattern.

That's why:

```python
re.fullmatch()
```

is useful.

Example:

```python
import re

pattern = r"\d+"

print(re.fullmatch(pattern, "123"))
print(re.fullmatch(pattern, "123abc"))
```

Output:

```text
<re.Match object ...>
None
```

The second string contains non-digit characters, so the entire string does not match.

# 62. `match()` vs `search()` vs `fullmatch()`

This distinction is important.

### `match()`

Checks the **beginning** of the string.

```python
re.match(pattern, text)
```

### `search()`

Searches **anywhere** in the string.

```python
re.search(pattern, text)
```

### `fullmatch()`

Requires the **entire string** to match.

```python
re.fullmatch(pattern, text)
```

Think of them as:

```text
match      → beginning
search     → anywhere
fullmatch  → everything
```

# 63. Common Beginner Mistakes

## Mistake 1: Forgetting That `.` Is Special

This:

```text
.
```

does not mean only a literal dot.

```text
\.
```

## Mistake 2: Forgetting `+`

Compare:

```text
\d
```

and:

```text
\d+
```

`\d` matches one digit.

`\d+` matches one or more digits.

For:

```text
12345
```

`\d` can match:

while:

```text
\d+
```

can match:

```text
12345
```

## Mistake 3: Using `search()` for Complete Validation

```python
re.fullmatch()
```

```python
re.search()
```

## Mistake 4: Forgetting Raw Strings

Regex patterns are often easier to write as:

```python
r"\d+"
```

instead of manually escaping backslashes.

## Mistake 5: Assuming a Simple Email Regex Is Perfect

A simple pattern such as:

```text
[\w.-]+@[\w.-]+\.\w+
```

is useful for learning and many basic applications.

However, real-world email syntax is considerably more complex.

Regex validation should therefore be treated as a **format check**, not proof that an email address actually exists.

# 66. Final Test Example

Try to understand this program before looking at the output:

```python
import re

text = """
My name is Kamraan.
My emails are kamraan@gmail.com and admin@example.org.
My numbers are 123 and 4567.
"""

email_pattern = re.compile(r"[\w.-]+@[\w.-]+\.\w+")
number_pattern = re.compile(r"\d+")

emails = email_pattern.findall(text)
numbers = number_pattern.findall(text)

print("Emails:", emails)
print("Numbers:", numbers)

cleaned_text = re.sub(r"\s+", " ", text).strip()

print("Cleaned text:")
print(cleaned_text)
```

### What happened?

First, we created a reusable email pattern:

```python
email_pattern = re.compile(r"[\w.-]+@[\w.-]+\.\w+")
```

Then:

```python
email_pattern.findall(text)
```

found all email-like patterns.

We created another pattern:

```python
number_pattern = re.compile(r"\d+")
```

which found all groups of digits.

Finally:

```python
re.sub(r"\s+", " ", text)
```

replaced repeated whitespace with a single space.
