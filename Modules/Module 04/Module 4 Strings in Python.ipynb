# 1. Slicing in Strings

String slicing means extracting a **part of a string**.

Consider:

```python
text = "Python"
```

The characters have indexes:

```text
 P   y   t   h   o   n
 0   1   2   3   4   5
```

Python indexing starts from:

```text
0
```

not `1`.

---

## Accessing a Single Character

```python
text = "Python"

print(text[0])
print(text[1])
print(text[2])
```

Output:

```text
P
y
t
```

---

## Negative Indexing

Python also allows negative indexes.

```text
 P   y   t   h   o   n
-6  -5  -4  -3  -2  -1
```

Example:

```python
text = "Python"

print(text[-1])
print(text[-2])
```

Output:

```text
n
o
```

The last character can therefore be accessed using:

```python
text[-1]
```

---

# 2. String Slicing Syntax

The basic syntax is:

```python
string[start:end]
```

The `start` index is included.

The `end` index is **not included**.

For example:

```python
text = "Python"

print(text[0:3])
```

Output:

```text
Pyt
```

Python takes indexes:

```text
0
1
2
```

but stops before index:

```text
3
```

---

## Example

```python
text = "Programming"

print(text[0:4])
```

Output:

```text
Prog
```

---

## Slicing from the Beginning

You can leave the start index empty.

```python
text = "Python"

print(text[:3])
```

Output:

```text
Pyt
```

This means:

```python
text[0:3]
```

---

## Slicing Until the End

You can leave the end index empty.

```python
text = "Python"

print(text[2:])
```

Output:

```text
thon
```

This means:

```python
text[2:6]
```

---

## Copying a String with Slicing

You can use:

```python
text = "Python"

copy_text = text[:]

print(copy_text)
```

Output:

```text
Python
```

---

# 3. Slicing with Step

The complete slicing syntax is:

```python
string[start:end:step]
```

Example:

```python
text = "Python"

print(text[0:6:2])
```

Output:

```text
Pto
```

Python takes:

```text
P
t
o
```

because the step is `2`.

---

## Step of 1

```python
text = "Python"

print(text[0:6:1])
```

Output:

```text
Python
```

---

## Step of 2

```python
text = "Python"

print(text[::2])
```

Output:

```text
Pto
```

This starts from the beginning and takes every second character.

---

## Step of 3

```python
text = "Programming"

print(text[::3])
```

Output:

```text
Pga
```

---

# 4. Reversing a String

A very useful slicing technique is:

```python
[::-1]
```

Example:

```python
text = "Python"

print(text[::-1])
```

Output:

```text
nohtyP
```

### Why?

A step of `-1` means:

> Move through the string from right to left.

---

## Example

```python
word = "Hello"

print(word[::-1])
```

Output:

```text
olleH
```

---

# 5. Practical Slicing Examples

```python
text = "Python Programming"

print(text[:6])
print(text[7:])
print(text[0:6])
print(text[::-1])
```

Output:

```text
Python
Programming
Python
gnimmargorP nohtyP
```

# 6. Important Point: Strings Are Immutable

Strings in Python are **immutable**.

This means that once a string is created, its individual characters cannot be changed directly.

For example:

```python
text = "Python"

text[0] = "J"
```

This produces an error.

You cannot directly replace the character at index `0`.

Instead, create a new string:

```python
text = "Python"

text = "J" + text[1:]

print(text)
```

Output:

```text
Jython
```

# 7. f-Strings in Python

An **f-string** is a convenient way to insert variables and expressions directly into a string.

The syntax is:

```python
f"Text {variable}"
```

Example:

```python
name = "Kamraan"
age = 25

print(f"My name is {name} and I am {age} years old.")
```

Output:

```text
My name is Kamraan and I am 25 years old.
```

---

## Why Use f-Strings?

Without an f-string:

```python
name = "Kamraan"
age = 25

print("My name is", name, "and I am", age, "years old.")
```

With an f-string:

```python
name = "Kamraan"
age = 25

print(f"My name is {name} and I am {age} years old.")
```

The f-string is usually easier to read.

---

# 8. Basic f-String Syntax

Put an `f` before the opening quotation mark:

```python
f"Hello {name}"
```

Example:

```python
name = "Rahul"

message = f"Hello {name}"

print(message)
```

Output:

```text
Hello Rahul
```

---

## Using Multiple Variables

```python
name = "Rahul"
age = 25
city = "Srinagar"

print(f"{name} is {age} years old and lives in {city}.")
```

Output:

```text
Rahul is 25 years old and lives in Srinagar.
```

# 32. String Slicing vs String Methods

These are two different ways of working with strings.

### Slicing

Used to extract part of a string:

```python
text = "Python"

print(text[0:3])
```

Output:

```text
Pyt
```

### Methods

Used to perform operations on a string:

```python
text = "python"

print(text.upper())
```

Output:

```text
PYTHON
```

A useful way to remember:

> **Slicing selects characters. Methods perform operations.**

---

# 33. Important Point: String Methods Return New Strings

Because strings are immutable, methods such as `upper()` and `replace()` do not modify the original string in place.

Example:

```python
text = "python"

text.upper()

print(text)
```

Output:

```text
python
```

The original string is unchanged.

To keep the result:

```python
text = "python"

text = text.upper()

print(text)
```

Output:

```text
PYTHON
```
