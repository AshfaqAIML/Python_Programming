# 1. Introduction to OOP

OOP stands for:

> **Object-Oriented Programming**

It is a way of designing programs around **objects** rather than only around individual functions and variables.

An object can contain:

* Data
* Functions that operate on that data

For example, a bank account object might contain:

```text
Account Number
Account Holder
Balance
```

and methods such as:

```text
deposit()
withdraw()
check_balance()
```

---

Imagine building a large application for a university.

There may be:

```text
Students
Teachers
Courses
Departments
Exams
Payments
```

```text
Student
 ├── name
 ├── age
 ├── course
 └── display_info()

Teacher
 ├── name
 ├── subject
 └── teach()
```

# 20. Alternative Constructor with `classmethod`

Class methods are often used to create alternative constructors.

Example:

```python id="x2m7q9"
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, data):

        return cls(name, int(age))

```

Output:

```text id="k8p3m6"
Kamraan
25
```

# 21. Static Methods

A static method does not automatically receive `self` or `cls`.

It uses:

```python id="q4x8m1"
@staticmethod
```

Example:

```python id="v6m2p9"
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b


print(Calculator.add(10, 20))
```

Output:

```text id="x3k7q5"
30
```

---

# 22. When to Use Static Methods

Use a static method when a function:

* Belongs logically to the class
* Does not need object-specific data
* Does not need class-specific data

Example:

```python id="m8q3v1"
class MathTools:

    @staticmethod
    def square(number):
        return number * number


print(MathTools.square(5))
```

Output:

```text id="p2x6k9"
25
```

---

# 23. Instance vs Class vs Static Methods

| Type            | First Parameter    | Works Mainly With         |
| --------------- | ------------------ | ------------------------- |
| Instance method | `self`             | Object data               |
| Class method    | `cls`              | Class data                |
| Static method   | None automatically | Independent utility logic |

A useful memory trick:

```text
self → object
cls  → class
static → neither automatically
```

Now:

```python id="v4m8x1"
phone_book = PhoneBook()

phone_book.add_contact(
    Contact("Rahul", "9876543210")
)

phone_book.add_contact(
    Contact("Aman", "9123456780")
)

contact = phone_book.search_contact("Rahul")

if contact:
    contact.show_contact()
else:
    print("Contact not found")
```

Output:

```text id="k7p3q9"
Rahul: 9876543210
```

# 32. Using `super()`

Instead of repeating the parent's initialization, we can use:

```python id="v7m2q8"
super()
```

Example:

```python id="q3x9p4"
class Animal:

    def __init__(self, name):
        self.name = name


class Dog(Animal):

    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed


dog = Dog("Bruno", "German Shepherd")

print(dog.name)
print(dog.breed)
```

Output:

```text id="k8m5x2"
Bruno
German Shepherd
```

`super()` allows the child class to call methods from its parent class.

---

# 33. Single Inheritance with Arguments

Parent and child classes can work with multiple arguments.

```python id="x6q1m9"
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):

    def __init__(self, name, age, course):
        super().__init__(name, age)
        self.course = course


student = Student("Kamraan", 25, "Python")

print(student.name)
print(student.age)
print(student.course)
```

# 38. Polymorphism: Operator Overloading

Python allows classes to define how operators behave for their objects.

This is called **operator overloading**.

For example:

```python id="x4m7q2"
class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(
            self.x + other.x,
            self.y + other.y
        )

    def show(self):
        print(self.x, self.y)


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2

p3.show()
```

Output:

```text id="n9q6m3"
6 8
```

---

```python id="s2m8x5"

```

Python can internally use:

```python id="q4x7m1"
p1.__add__(p2)
```

We defined that behavior ourselves.

This allows our custom objects to work naturally with operators.

# 41. Polymorphism: Method Overloading

In some languages, method overloading means creating multiple methods with the same name but different parameter lists.

Python does **not** support traditional method overloading in the same way as languages such as Java or C++.

For example:

```python id="v1m6q8"
class Calculator:

    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c
```

The second `add()` replaces the first one.

Calling:

```python id="j3x9p4"
calculator.add(10, 20)
```

would not work as a traditional overloaded method.

We may want outside code to use:

```python
account.deposit(500)
```

instead of directly changing:

```python
account.balance = -100000
```

Encapsulation helps protect object data from inappropriate use.

# 51. Abstract Classes

An **abstract class** is a class designed to be inherited from rather than instantiated directly.

This will raise an error:

Balance: ₹6000

# 63. Common Beginner Mistakes

## Mistake 1: Forgetting `self`

Incorrect:

```python id="m9x4q7"
class Student:

    def greet():
        print("Hello")
```

Correct:

```python id="p3q8m1"
class Student:

    def greet(self):
        print("Hello")
```

---

## Mistake 2: Forgetting `self.`

Incorrect:

```python id="x5m7q2"
class Student:

    def __init__(self, name):
        name = name
```

Correct:

```python id="n8p3m6"
class Student:

    def __init__(self, name):
        self.name = name
```

Without `self.name`, the value is not stored as an instance attribute.

---

## Mistake 3: Confusing `self` and `cls`

Remember:

```text id="v2q7m4"
self → current object
cls  → current class
```

---

## Mistake 4: Forgetting `super()`

When the child class has its own initializer and also needs the parent's initialization:

```python id="q6m3x8"
super().__init__(...)
```

can be used.

---

## Mistake 5: Expecting Traditional Method Overloading

Python does not keep multiple methods with the same name based only on parameter count.

This:

```python id="m8x2q5"
def add(self, a, b):
    ...


def add(self, a, b, c):
    ...
```

causes the second definition to replace the first.

Use default values or `*args` when appropriate.

---

## Mistake 6: Directly Modifying Internal Data

Instead of:

```python id="x4q8m1"
account.__balance = -500
```

design methods such as:

```python id="p7m3x9"
account.deposit(500)
account.withdraw(200)
```

to control valid operations.

---

# 64. Short Summary

A **class** is a blueprint:

```python id="n3x8m5"
class Student:
    pass
```
