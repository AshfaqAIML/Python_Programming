# 10. Associativity and Precedence of Operators

When an expression contains multiple operators, Python needs rules to determine which operation should happen first.

These rules are called **operator precedence**.

Associativity determines the direction in which operators with the same precedence are evaluated.

---

## Example

Consider:

```python
result = 10 + 5 * 2

print(result)
```

Output:

```text
20
```

Why not `30`?

Because multiplication has higher precedence than addition.

Python evaluates:

```text
5 * 2 = 10
```

Then:

```text
10 + 10 = 20
```

---

## Parentheses Have High Priority

```python
result = (10 + 5) * 2

print(result)
```

Output:

```text
30
```

The parentheses are evaluated first.

---

## Common Precedence Order

A simplified order is:

```text
1. Parentheses
2. Exponentiation **
3. Unary + and -
4. *, /, //, %
5. +, -
6. Comparisons
7. not
8. and
9. or
```

For example:

```python
result = 10 + 2 * 3
```

First:

```text
2 * 3 = 6
```

Then:

```text
10 + 6 = 16
```

So:

```python
print(result)
```

Output:

```text
16
```

---

## Associativity

Associativity determines how operators of the same precedence are grouped.

For many arithmetic operators, evaluation occurs from left to right.

Example:

```python
result = 20 - 5 - 3

print(result)
```

Output:

```text
12
```

Python evaluates:

```text
20 - 5 = 15
15 - 3 = 12
```

---

## Exponentiation Is Special

Exponentiation associates from right to left.

For example:

```python
result = 2 ** 3 ** 2

print(result)
```

Python interprets this as:

```text
2 ** (3 ** 2)
```

First:

```text
3 ** 2 = 9
```

Then:

```text
2 ** 9 = 512
```

Output:

```text
512
```

---

## Best Practice

When an expression becomes difficult to understand, use parentheses.

Instead of:

```python
result = a + b * c - d
```

you can make the intention clearer:

```python
result = a + (b * c) - d
```

Parentheses improve readability and reduce mistakes.

The following program combines many concepts from this module:
