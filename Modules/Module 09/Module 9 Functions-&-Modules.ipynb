## Positional Arguments

Arguments are matched according to their position.

Here:

```text id="a2k8z5"

```

---

## Order Matters

```python id="e7q1w9"
def divide(a, b):
    return a / b

print(divide(10, 2))
```

Output:

```text id="x3m6p1"
5.0
```

But:

```python id="s8k2v4"
print(divide(2, 10))
```

Output:

```text id="c5n9r7"
0.2
```

The order changes the result.

---

# 5. Keyword Arguments

With keyword arguments, you explicitly specify the parameter name.

```python id="z1p6m8"
def introduce(name, age):
    print(f"Name: {name}")
    print(f"Age: {age}")

introduce(age=25, name="Kamraan")
```


# 7. Variable-Length Positional Arguments

Sometimes we do not know how many positional arguments a function will receive.

Python provides:

```python id="r8m3v6"
*args
```

Example:

```python id="p2k7x9"
def add(*args):
    total = 0

    for number in args:
        total += number

    return total

print(add(10, 20))
print(add(10, 20, 30))
print(add(1, 2, 3, 4, 5))
```

Output:

```text id="n6q4w8"
30
60
15
```

---

## What Is `args`?

Inside the function, `args` is a tuple containing all positional arguments.

Example:

```python id="y5m1p7"
def show(*args):
    print(args)

show(10, 20, 30)
```

Output:

```text id="c8r3v2"
(10, 20, 30)
```

The name `args` is conventional.

You could technically use another name:

```python id="k4x9n1"
def show(*numbers):
    print(numbers)
```

But `*args` is the common Python convention.

---

# 8. Variable-Length Keyword Arguments

For an unknown number of keyword arguments, Python provides:

```python id="q7m2v5"
**kwargs
```

Example:

```python id="s3x8n6"
def show_info(**kwargs):
    print(kwargs)

show_info(name="Kamraan", age=25, city="Srinagar")
```

Output:

```text id="w1p6k9"
{'name': 'Kamraan', 'age': 25, 'city': 'Srinagar'}
```

Inside the function, `kwargs` is a dictionary.

---

## Looping Through `kwargs`

```python id="a9v4m2"
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_info(
    name="Kamraan",
    age=25,
    city="Srinagar"
)
```

Output:

```text id="z6k3p8"
name: Kamraan
age: 25
city: Srinagar
```

---

# 9. `*args` vs `**kwargs`

| Feature         | `*args`              | `**kwargs`                   |
| --------------- | -------------------- | ---------------------------- |
| Handles         | Positional arguments | Keyword arguments            |
| Inside function | Tuple                | Dictionary                   |
| Example         | `add(10, 20, 30)`    | `user(name="Rahul", age=25)` |

Example:

```python id="m5x8q2"
def example(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

example(10, 20, name="Rahul", age=25)
```

Output:

```text id="r2n7v4"
Args: (10, 20)
Kwargs: {'name': 'Rahul', 'age': 25}
```

---

# 10. Docstrings in Python Functions

A **docstring** is a string used to document a function, class, or module.

It is normally written immediately after the function definition.

Example:

```python id="x4p8m1"
def add(a, b):
    """Return the sum of two numbers."""
    return a + b
```

Output:

```text id="r8k4v1"
5
4
3
2
1
Done
```

---

## How Recursion Works

The function keeps calling itself:

```text id="c7m2x5"
countdown(5)
      ↓
countdown(4)
      ↓
countdown(3)
      ↓
countdown(2)
      ↓
countdown(1)
      ↓
countdown(0)
```

At `0`, the stopping condition is reached.

---

# 12. Base Case in Recursion

Every useful recursive function needs a **base case**.

The base case tells the function when to stop calling itself.

Example:

```python id="m3q7v8"
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```

Output:

```text id="z1k5p9"
120
```

The base case is:

```python id="h8x2m4"
if n == 0:
    return 1
```

---

## Factorial Explanation

Mathematically:

```text id="2f7q9m"
5! = 5 × 4 × 3 × 2 × 1
```

The recursive process is:

```text id="j3m8x1"
5 × factorial(4)
       ↓
4 × factorial(3)
       ↓
3 × factorial(2)
       ↓
2 × factorial(1)
       ↓
1 × factorial(0)
       ↓
1
```

The final result is:

```text id="x6p2r7"
120
```

# 13. Local and Global Variables

Variables can have different scopes.

Two important scopes are:

* Local
* Global

---

## Local Variable

A variable created inside a function is normally local to that function.

```python id="v5n8q3"
def greet():
    message = "Hello"

    print(message)

greet()
```

Output:

```text id="m7x2k9"
Hello
```

The variable:

```text id="s4p1v6"
message
```

belongs to the function.

Trying to use it outside:

```python id="q8m3x5"
print(message)
```

will cause an error because `message` is not defined in that outer scope.

---

## Global Variable

A variable defined outside a function is global.

```python id="y2k7m4"
name = "Kamraan"

def greet():
    print(name)

greet()
```

Output:

```text id="f9x3p8"
Kamraan
```

The function can read the global variable.

---

# 14. Modifying a Global Variable

If you want to modify a global variable from inside a function, use the `global` keyword.

```python id="n6m1q8"
count = 0

def increase():
    global count
    count += 1

increase()
increase()

print(count)
```

Output:

```text id="p4x7k2"
2
```

### Important

Use global variables carefully.

Too many global variables can make large programs harder to understand and maintain.

Example:


This is similar to:



Output:

```text id="p7v2k5"
[1, 4, 9, 16, 25]
```

### How does it work?

The lambda function:

```python id="y6n3m8"
lambda x: x * x
```

is applied to every number.

```text id="3r8q2p"
1 → 1
2 → 4
3 → 9
4 → 16
5 → 25
```

---

# 20. `filter()` with Lambda

The `filter()` function keeps only the items that satisfy a condition.

Example:

```python id="k5x9m2"
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(
    filter(lambda x: x % 2 == 0, numbers)
)

print(even_numbers)
```

Output:

```text id="v1q7p4"
[2, 4, 6]
```

The lambda checks:

```python id="z8m3x6"
x % 2 == 0
```

Only numbers that satisfy this condition are kept.

---

# 21. `map()` vs `filter()`

| Function   | Purpose                             |
| ---------- | ----------------------------------- |
| `map()`    | Transform every item                |
| `filter()` | Keep items that satisfy a condition |

Example:

```python id="w2n6q8"
numbers = [1, 2, 3, 4]

squares = list(map(lambda x: x * x, numbers))

even = list(filter(lambda x: x % 2 == 0, numbers))
```

Results:

```text id="j4p8m1"
squares → [1, 4, 9, 16]
even    → [2, 4]
```

Its value depends on how the Python file is being used.

When a file is run directly, Python sets:

```python id="w4p7n2"

```

When the file is imported as a module, its `__name__` normally becomes the module's name.

---

# 29. Basic `__name__` Example

Create:

```text id="m6x2q9"
test.py
```

When this file is run directly, `greet()` executes.

But when the file is imported into another Python file, the code inside:

```python id="n7p3v9"

```

does not execute automatically.

---

# 31. Why Is `__name__` Useful?

Suppose we have:

```python id="k2x6m8"
# calculator.py

def add(a, b):
    return a + b


if __name__ == "__main__":
    print(add(10, 20))
```

```text id="q4m9p1"

```

But if another file imports it:

```python id="w7n3x5"
import calculator

print(calculator.add(5, 10))
```

Output:

```text id="v8p2k6"
15
```

The test code inside the `__main__` block does not run during import.

---

# 32. Simple Banking Application

Let's combine functions, variables, conditions, loops, and user input to create a small banking application.

The application will support:

* Checking balance
* Depositing money
* Withdrawing money
* Exiting the program

---

## Step 1: Create the Balance

```python id="m5x8q2"
balance = 1000
```

### `while` Loop

# 38. Common Beginner Mistakes

## Mistake 1: Forgetting to Call the Function

Defining:

```python id="p8m3x7"
def greet():
    print("Hello")
```

does not execute it.

You need:

```python id="v4q9k2"
greet()
```

## Mistake 2: Forgetting `return`

Incorrect:

```python id="m7x2p8"
def add(a, b):
    a + b
```

Correct:

```python id="q3n8v5"
def add(a, b):
    return a + b
```

---

## Mistake 3: Confusing `*args` and `**kwargs`

Remember:

```text id="f2k7m9"
*args   → positional arguments → tuple
```

This never reaches a stopping condition.

Correct:

```python id="n5q2v7"
def count(n):
    if n == 0:
        return

    print(n)
    count(n - 1)
```

---

## Mistake 5: Calling a Function While Passing It

If another function expects a function:

```python id="r7m3x9"
execute(greet)
```

is different from:

```python id="k2q8v5"
execute(greet())
```

The first passes the function.

The second calls it immediately and passes its result.

---

## Mistake 6: Importing a Module Incorrectly

If you have:

```text id="v6m1p8"
calculator.py
```

and want to use it:

```python id="q4x7n2"

```

Then call:

```python id="s9m3k5"
calculator.add(10, 20)
```

# 39. Short Summary

A **function** is a reusable block of code:

```python id="m3x7q9"
def greet():
    print("Hello")
```

A function can receive parameters:

```python id="p8n2v6"
def greet(name):
    print(f"Hello {name}")
```

A function can return a value:

```python id="x4m9k1"
def add(a, b):
    return a + b
```

Arguments can be passed positionally or by keyword:

```python id="q7m3v5"
add(10, 20)
```

or:

```python id="n8x2p6"
add(a=10, b=20)
```

`*args` handles variable positional arguments:

```python id="m5q9v1"
def add(*args):
    ...
```

`**kwargs` handles variable keyword arguments:

```python id="r3x7k8"
def user(**kwargs):
    ...
```

Docstrings document functions:

```python id="p6m2v9"
def add(a, b):
    """Return the sum of two numbers."""
    return a + b
```

Recursive functions call themselves:

```python id="x8q4m1"
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n - 1)
```

Lambda functions provide short anonymous functions:

```python id="v2m7k5"
square = lambda x: x * x
```

`map()` transforms values:

```python id="n4x8p2"
map(lambda x: x * 2, numbers)
```

`filter()` selects values:

```python id="q6m3v9"
filter(lambda x: x % 2 == 0, numbers)
```

A module is a Python file containing reusable code:

```python id="p7x2m5"
import math
```

A user-defined module can be created as a normal `.py` file.

Finally:

```python id="k3q8v1"
if __name__ == "__main__":
```

allows code to run only when the file is executed directly.

Functions can accept different types of arguments:

```python id="v7p2m5"
positional
keyword
default
*args
**kwargs
```

They can also:

```text id="n3x9k6"
Return values
    ↓
Call other functions
    ↓
Be passed as arguments
    ↓
Call themselves recursively
```

Lambda functions provide a concise way to create small functions, while `map()` and `filter()` make it convenient to transform and select data.

Modules take code organization one step further by allowing related functions and classes to be stored in separate Python files.

The overall idea is:

> **Functions organize behavior, while modules organize code.**

Once you understand functions and modules, your Python programs can become much more **organized, reusable, readable, and maintainable**.
