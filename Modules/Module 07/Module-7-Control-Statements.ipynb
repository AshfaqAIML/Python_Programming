# Control Statements in Python

Control statements are used to **control the flow of execution** in a Python program.

Normally, Python executes code from top to bottom.

But sometimes we want the program to:

* Make decisions
* Execute code only when a condition is true
* Choose between multiple conditions
* Check conditions inside other conditions
* Write simple conditions in one line

This notebook explains:

* The `if` block
* The `if-else` block
* The `if-elif-else` statement
* Nested `if` statements
* The ternary operator

---

# 1. The `if` Block

The `if` statement is used when we want to execute some code **only when a condition is true**.

The basic syntax is:

```python
if condition:
    statement
```

The condition must produce either:

```python
True
```

or:

```python
False
```

---

## Example of `if`

```python
age = 20

if age >= 18:
    print("You are an adult")
```

Output:

```python
You are an adult
```

### Why did the message print?

The condition is:

```python
age >= 18
```

Since:

```python
20 >= 18
```

is `True`, Python executes the `print()` statement.

---

## Example When the Condition Is False

```python
age = 15

if age >= 18:
    print("You are an adult")
```

Output:

```python
```

Nothing is printed.

### Why?

Because:

```python
15 >= 18
```

is `False`.

Therefore, Python skips the code inside the `if` block.

---

## 2. Structure of an `if` Block

```python
if condition:
    # code to execute
```

For example:

```python
temperature = 35

if temperature > 30:
    print("It is hot")
```

Output:

```python
It is hot
```

The colon `:` is important.

Python uses indentation to determine which statements belong to the `if` block.

---

## Indentation in `if`

Correct:

```python
age = 20

if age >= 18:
    print("Adult")
```

Incorrect:

```python
age = 20

if age >= 18:
print("Adult")
```

Python will produce an indentation error.

### Important Rule

Code inside an `if` statement must be indented.

Usually, Python uses **4 spaces**.

---

# 2. `if-else` Block

Sometimes we want the program to do one thing when a condition is true and another thing when it is false.

For this, we use:

```python
if condition:
    statement1
else:
    statement2
```

The `if` block runs when the condition is `True`.

The `else` block runs when the condition is `False`.

---

## Example of `if-else`

```python
age = 20

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")
```

Output:

```python
You can vote
```

---

## Example When Condition Is False

```python
age = 15

if age >= 18:
    print("You can vote")
else:
    print("You cannot vote")
```

Output:

```python
You cannot vote
```

### Why?

The condition:

```python
age >= 18
```

is false because:

```python
15 >= 18
```

is `False`.

Therefore, Python executes the `else` block.

---

# 4. `if-elif-else`

Sometimes we have more than two possible conditions.

For example, we may want to classify marks:

* 90 or above → A
* 80 or above → B
* 70 or above → C
* Otherwise → D

For this situation, we use:

```python
if condition1:
    statement
elif condition2:
    statement
elif condition3:
    statement
else:
    statement
```

`elif` means:

> else if

---

## Example of `if-elif-else`

```python
marks = 85

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
else:
    print("Grade D")
```

Output:

```python
Grade B
```

### Why?

Python checks the conditions from top to bottom.

First:

```python
marks >= 90
```

is false.

Then:

```python
marks >= 80
```

is true.

So Python prints:

```python
Grade B
```

After finding a true condition, Python skips the remaining `elif` and `else` blocks.

---

## 5. Order of Conditions Matters

Consider:

```python
marks = 95

if marks >= 70:
    print("Grade C")
elif marks >= 80:
    print("Grade B")
elif marks >= 90:
    print("Grade A")
```

Output:

```python
Grade C
```

### Why?

The first condition:

```python
marks >= 70
```

is already true.

Python stops checking further conditions.

Therefore, the order should normally go from the **most specific/highest condition to the lower conditions**.

Correct:

```python
marks = 95

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
else:
    print("Grade D")
```

Output:

```python
Grade A
```

# 7. Nested `if` Statements

A **nested `if` statement** means placing one `if` statement inside another `if` statement.

In simple words:

> An `if` inside another `if`.

Example:

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
```

Output:

```python
Entry allowed
```

---

## How Nested `if` Works

The outer condition is checked first:

```python
age >= 18
```

If it is true, Python checks the inner condition:

```python
has_id
```

Only when both conditions are true does the message print.

---

## Example When Inner Condition Is False

```python
age = 20
has_id = False

if age >= 18:
    if has_id:
        print("Entry allowed")
```

Output:

```python
```

### Why?

The outer condition is true:

```python
20 >= 18
```

But the inner condition is false:

```python
has_id = False
```

Therefore, the `print()` statement is not executed.

---

# 8. Nested `if` with `else`

We can also use `else` inside a nested `if`.

```python
age = 20
has_id = False

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("You must be 18 or older")
```

Output:

```python
ID required
```

### How does it work?

First Python checks:

```python
age >= 18
```

This is true.

Then Python checks:

```python
has_id
```

This is false.

Therefore, the inner `else` executes.

---

# 9. Real-Life Example of Nested `if`

Imagine an online exam system.

```python
registered = True
```

Output:


The second condition is checked only after the first condition succeeds.

# 10. Nested `if` vs `if-elif-else`

These two structures have different purposes.

### Nested `if`

Used when one condition depends on another condition.

```python
if condition1:
    if condition2:
        print("Both conditions are true")
```

### `if-elif-else`

Used when choosing between different alternatives.

```python
if condition1:
    print("Option 1")
elif condition2:
    print("Option 2")
else:
    print("Option 3")
```

---

# 11. The Ternary Operator

The **ternary operator** is a short way of writing a simple `if-else` statement in one line.

Normal `if-else`:

```python
age = 20

if age >= 18:
    result = "Adult"
else:
    result = "Minor"
```

We can write the same thing using a ternary expression:

```python
age = 20

result = "Adult" if age >= 18 else "Minor"

print(result)
```

Output:

```python
Adult
```

---

## 12. Syntax of the Ternary Operator

The general syntax is:

```python
value_if_true if condition else value_if_false
```

For example:

```python
result = "Pass" if marks >= 40 else "Fail"
```

The condition:

```python
marks >= 40
```

is checked first.

If it is true:

```python
"Pass"
```

is assigned.

Otherwise:

```python
"Fail"
```

is assigned.

---

## Example of Ternary Operator

```python
marks = 75

result = "Pass" if marks >= 40 else "Fail"

print(result)
```

Output:

```python
Pass
```

---

## Example with Numbers

```python
a = 10
b = 20

larger = a if a > b else b

print(larger)
```

Output:

```python
20
```

### Why?

The condition is:

```python
a > b
```

which is:

```python
10 > 20
```

This is false.

Therefore, Python chooses:

```python
b
```

---

# 13. Ternary Operator with `print()`

You can also use a ternary expression directly inside `print()`.

```python
age = 17

print("Adult" if age >= 18 else "Minor")
```

Output:

```python
Minor
```

This is useful for very small decisions.

---

# 14. Ternary Operator vs Normal `if-else`

Normal `if-else`:

```python
age = 20

if age >= 18:
    status = "Adult"
else:
    status = "Minor"

print(status)
```

Ternary:

```python
age = 20

status = "Adult" if age >= 18 else "Minor"

print(status)
```

Both produce:

```python
Adult
```

### Which one should you use?

Use the ternary operator when the condition is **short and easy to understand**.

Use normal `if-else` when the logic is more complicated.

---

# 15. Avoid Complicated Ternary Expressions

This is difficult to read:

```python
result = "A" if marks >= 90 else "B" if marks >= 80 else "C"
```

Although it works, a normal `if-elif-else` is easier to understand:

```python
if marks >= 90:
    result = "A"
elif marks >= 80:
    result = "B"
else:
    result = "C"
```

### Good Programming Practice

> Use the ternary operator for simple decisions, not complicated logic.

# 23. Common Mistakes

## Mistake 1: Forgetting the Colon

Incorrect:

```python
if age >= 18
    print("Adult")
```

Correct:

```python
if age >= 18:
    print("Adult")
```

---

## Mistake 2: Incorrect Indentation

Incorrect:

```python
if age >= 18:
print("Adult")
```

Correct:

```python
if age >= 18:
    print("Adult")
```

---

## Mistake 3: Using `=` Instead of `==`

Remember:

```python
=
```

means assignment.

While:

```python
==
```

means comparison.

Correct:

```python
age = 20

if age == 20:
    print("Age is 20")
```

---

## Mistake 4: Using Multiple Independent `if` Statements When `elif` Is Needed

Consider:

```python
marks = 95

if marks >= 70:
    print("C")

if marks >= 80:
    print("B")

if marks >= 90:
    print("A")
```

Output:

```python
C
B
A
```

All three conditions are checked.

If you want **only one result**, use `if-elif-else`:

```python
marks = 95

if marks >= 90:
    print("A")
elif marks >= 80:
    print("B")
elif marks >= 70:
    print("C")
```

Output:

```python
A
```

# 26. Short Summary

The `if` statement is used to make a decision:

```python
if condition:
    code
```

The `if-else` statement provides two choices:

```python
if condition:
    code1
else:
    code2
```

The `if-elif-else` statement provides multiple choices:

```python
if condition1:
    code1
elif condition2:
    code2
else:
```

```python
if condition1:
```

```python
result = value1 if condition else value2
```
