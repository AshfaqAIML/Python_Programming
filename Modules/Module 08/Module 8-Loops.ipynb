# 1. `for` Loop

A `for` loop is used to repeat code for each item in a sequence or iterable.

The basic syntax is:

```python
for variable in sequence:
    code
```

Example:

```python
for number in [1, 2, 3, 4, 5]:
    print(number)
```

Output:

```text
1
2
3
4
5
```

### How does it work?

Python takes one item at a time from the list.

```text
1 → print
2 → print
3 → print
4 → print
5 → print
```

The loop automatically stops after all items have been processed.

---

## Simple Example

```python
for i in [10, 20, 30]:
    print(i)
```

Output:

```text
10
20
30
```

Here, `i` is the loop variable.

# 8. Highest Value Using a Loop

We can find the highest number manually using a loop.

```python
numbers = [10, 50, 30, 80, 20]

highest = numbers[0]

for number in numbers:
    if number > highest:
        highest = number

print("Highest:", highest)
```

Output:

```text
Highest: 80
```

### Why?

We start by assuming the first number is the highest.

Then we compare every other number with it.

Whenever we find a larger number, we update `highest`.

---

# 9. Lowest Value Using a Loop

The same idea can be used to find the smallest number.

```python
numbers = [10, 50, 30, 80, 20]

lowest = numbers[0]

for number in numbers:
    if number < lowest:
        lowest = number

print("Lowest:", lowest)
```

Output:

```text
Lowest: 10
```

---

# 10. Total, Highest and Lowest Together

```python
numbers = [10, 50, 30, 80, 20]

total = 0
highest = numbers[0]
lowest = numbers[0]

for number in numbers:
    total += number

    if number > highest:
        highest = number

    if number < lowest:
        lowest = number

print("Total:", total)
print("Highest:", highest)
print("Lowest:", lowest)
```

Output:

```text
Total: 190
Highest: 80
Lowest: 10
```

This demonstrates how a single loop can perform multiple operations.

# 12. `continue`

The `continue` statement skips the current iteration and moves to the next iteration.

Example:

```python
for number in range(1, 6):
    if number == 3:
        continue

    print(number)
```

Output:

```text
1
2
4
5
```

### Why?

When `number` becomes `3`, Python skips:

```python
print(number)
```

and continues with the next iteration.

# 15. The `while` Loop

A `while` loop repeats code **as long as a condition is true**.

The basic syntax is:

```python
while condition:
    code
```

Example:

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

Output:

```text
1
2
3
4
5
```

---

## How Does a `while` Loop Work?

Python checks the condition:

```text
count <= 5
```

If it is true, the loop runs.

After each iteration:

```python
count += 1
```

increases the value.

Eventually:

```text
count = 6
```

Then:

```text
6 <= 5
```

is false, so the loop stops.

---

# 16. `for` Loop vs `while` Loop

A `for` loop is commonly used when you know what you want to iterate over.

```python
for number in range(1, 6):
    print(number)
```

A `while` loop is useful when repetition depends on a condition.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

| `for` Loop                        | `while` Loop                                   |
| --------------------------------- | ---------------------------------------------- |
| Iterates over a sequence/iterable | Runs while a condition is true                 |
| Often used for known iterations   | Useful for condition-based repetition          |
| Usually easier for collections    | Useful when stopping depends on changing state |

# 22. Nested Loops

A **nested loop** means putting one loop inside another loop.

Example:

```python
for i in range(3):
    for j in range(3):
        print(i, j)
```

Output:

```text
0 0
0 1
0 2
1 0
1 1
1 2
2 0
2 1
2 2
```

### How does it work?

For every iteration of the outer loop, the inner loop runs completely.

Think of it as:

```text
Outer loop
    ↓
Inner loop runs completely
    ↓
Outer loop moves to next value
    ↓
Inner loop runs completely again
```

---

The outer loop creates three rows.

# 29. Number Guessing Game — Problem

Let's design a simple number guessing game.

The computer should:

1. Generate a random number.
2. Ask the player to guess.
3. Tell the player whether the guess is too high or too low.
4. Continue until the player guesses correctly.
5. Count the number of attempts.

The program will need:

```text
random
while loop
input()
int()
if-elif-else
break
```

---

# 30. Number Guessing Game — Basic Version

First, generate a random number:

```python
import random

secret_number = random.randint(1, 100)

print("I have chosen a number between 1 and 100.")
```

Now we need to repeatedly ask for guesses.

---

# 31. Number Guessing Game — Solution

```python
import random

secret_number = random.randint(1, 100)
attempts = 0

print("Guess the number between 1 and 100.")

while True:
    guess = int(input("Enter your guess: "))
    attempts += 1

    if guess < secret_number:
        print("Too low!")

    elif guess > secret_number:
        print("Too high!")

    else:
        print("Correct!")
        print(f"You guessed it in {attempts} attempts.")
        break
```

Example interaction:

```text
Guess the number between 1 and 100.
Enter your guess: 50
Too high!
Enter your guess: 25
Too low!
Enter your guess: 37
Too high!
Enter your guess: 32
Correct!
You guessed it in 4 attempts.
```

The exact interaction changes because the secret number is randomly generated.

---

# 32. How the Guessing Game Works

First:

```python
secret_number = random.randint(1, 100)
```

creates a random number.

Then:

```python
while True:
```

creates a loop that continues until we explicitly stop it.

The player's guess is converted to an integer:

```python
guess = int(input("Enter your guess: "))
```

The attempt counter increases:

```python
attempts += 1
```

Then Python compares the guess.

If:

```python
guess < secret_number
```

the guess is too low.

If:

```python
guess > secret_number
```

the guess is too high.

Otherwise, the guess is correct.

Finally:

```python
break
```

stops the loop.

---

# 33. Improved Guessing Game

We can also limit the number of attempts.

```python
import random

secret_number = random.randint(1, 100)
max_attempts = 7

print("Guess the number between 1 and 100.")
print(f"You have {max_attempts} attempts.")

for attempt in range(1, max_attempts + 1):
    guess = int(input("Enter your guess: "))

    if guess < secret_number:
        print("Too low!")

    elif guess > secret_number:
        print("Too high!")

    else:
        print("Correct!")
        print(f"You won in {attempt} attempts.")
        break

else:
    print("You ran out of attempts.")
    print(f"The number was {secret_number}.")
```

### Important Point

The `else` belongs to the `for` loop.

A loop's `else` block runs when the loop finishes normally without encountering `break`.
