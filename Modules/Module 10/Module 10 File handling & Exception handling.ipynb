# 17. The `with` Statement

# 33. `else` with Exception Handling

Python allows an `else` block with `try-except`.

The `else` block runs when no exception occurs.

Example:

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid number.")

else:
    print("You entered:", number)
```

If the user enters:

```text
25
```

Output:

```text
You entered: 25
```

If the user enters:

```text
hello
```

Output:

```text
Invalid number.
```

The `else` block does not run in the second case.

---

# 34. `finally`

The `finally` block runs whether an exception occurs or not.

Example:

```python
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Invalid input.")

finally:
    print("Program finished.")
```

If the input is valid:

```text
Enter a number: 25
25
Program finished.
```

If the input is invalid:

```text
Enter a number: hello
Invalid input.
Program finished.
```

---

# 35. `try-except-else-finally`

The complete structure is:

```python
try:
    # code that may raise an exception

except SomeError:
    # handle exception

else:
    # runs if no exception occurs

finally:
    # always runs
```

Think of it as:

```text
try
 ↓
Did an error occur?
 ↓
Yes → except
 ↓
No → else
 ↓
finally → always
```

---

# 36. Practical Example

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)

finally:
    print("Calculation complete.")
```

# 38. Raising an Exception Inside a Function

```python
def calculate_square_root(number):
    if number < 0:
        raise ValueError("Number cannot be negative.")

    return number ** 0.5


print(calculate_square_root(25))
```

Output:

```text
5.0
```

```python
print(calculate_square_root(-25))
```

# 39. Catching a Raised Exception

We can raise and handle an exception.

```python
def withdraw(balance, amount):
    if amount <= 0:
        raise ValueError("Withdrawal amount must be positive.")

    if amount > balance:
        raise ValueError("Insufficient balance.")

    return balance - amount


try:
    balance = withdraw(1000, 1500)
    print(balance)

except ValueError as error:
    print("Error:", error)
```

Output:

```text
Error: Insufficient balance.
```

---

# 40. Why Raise Exceptions?

Raising exceptions is useful when a function receives invalid data.

For example:

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")

    return age
```

This prevents invalid values from silently entering the program.

# 52. Pickle vs JSON

Both JSON and pickle can store data, but they are designed for different purposes.

| JSON                                       | Pickle                               |
| ------------------------------------------ | ------------------------------------ |
| Text-based                                 | Binary                               |
| Widely supported                           | Mainly Python-specific               |
| Good for APIs and data exchange            | Good for Python object serialization |
| Human-readable                             | Usually not human-readable           |
| Safer for exchanging ordinary data formats | Must only load trusted pickle data   |

A very important security rule:

> **Never unpickle data from an untrusted or unknown source.**

Loading a malicious pickle can execute harmful code.

# 53. Pickle and Exception Handling

File problems can occur when working with pickle.

```python
import pickle

try:
    with open("student.pkl", "rb") as file:
        student = pickle.load(file)

    print(student)

except FileNotFoundError:
    print("Pickle file not found.")

except pickle.UnpicklingError:
    print("Could not load the pickle data.")

except EOFError:
    print("The pickle file is empty or incomplete.")
```

# 54. Complete Example: Save and Load Student Data

Let's combine:

* Functions
* Files
* JSON
* Exception handling

```python
import json


def save_student(student):
    with open("student.json", "w") as file:
        json.dump(student, file, indent=4)


def load_student():
    try:
        with open("student.json", "r") as file:
            return json.load(file)

    except FileNotFoundError:
        print("Student file does not exist.")
        return None

    except json.JSONDecodeError:
        print("Student file contains invalid JSON.")
        return None


student = {
    "name": "Kamraan",
    "age": 25,
    "course": "Python"
}

save_student(student)

loaded_student = load_student()

if loaded_student:
    print(loaded_student)
```

Possible output:

```text
{'name': 'Kamraan', 'age': 25, 'course': 'Python'}
```

# 55. Complete Example: Safe File Reader

A useful reusable function is a safe file reader.

```python
def read_file(filename):
    try:
        with open(filename, "r") as file:
            return file.read()

    except FileNotFoundError:
        print(f"File '{filename}' was not found.")
        return None

    except PermissionError:
        print(f"Permission denied for '{filename}'.")
        return None


content = read_file("notes.txt")

if content is not None:
    print(content)
```

This function handles common file errors without crashing the program.

# 56. Complete Example: Simple Text File Manager

We can combine file operations into a small program.

```python
def write_file(filename, content):
    with open(filename, "w") as file:
        file.write(content)


def read_file(filename):
    try:
        with open(filename, "r") as file:
            return file.read()

    except FileNotFoundError:
        return "File does not exist."


def append_file(filename, content):
    with open(filename, "a") as file:
        file.write(content)


write_file("notes.txt", "Python is easy.\n")

append_file("notes.txt", "Python is powerful.\n")

print(read_file("notes.txt"))
```

Output:

```text
Python is easy.
Python is powerful.
```

# 62. Final Test Example

Try to understand this complete program before looking at the output:

```python
import json


def save_data(filename, data):
    try:
        with open(filename, "w") as file:
            json.dump(data, file, indent=4)

        print("Data saved successfully.")

    except PermissionError:
        print("Permission denied.")


def load_data(filename):
    try:
        with open(filename, "r") as file:
            return json.load(file)

    except FileNotFoundError:
        print("File not found.")
        return None

    except json.JSONDecodeError:
        print("Invalid JSON data.")
        return None


student = {
    "name": "Kamraan",
    "course": "Python",
    "marks": [80, 90, 85]
}

save_data("student.json", student)

data = load_data("student.json")

if data is not None:
    print("Student:", data["name"])
    print("Course:", data["course"])
    print("Marks:", data["marks"])
```

Output:

```text
Data saved successfully.
Student: Kamraan
Course: Python
Marks: [80, 90, 85]
```

File handling allows Python programs to **store and retrieve information outside the running program**.

The basic pattern is:

```python
with open("file.txt", "r") as file:
    content = file.read()
```

For writing:

```python
with open("file.txt", "w") as file:
    file.write("Hello Python")
```

```python
with open("file.txt", "a") as file:
    file.write("\nNew content")
```

Exception handling allows programs to respond to unexpected situations instead of crashing immediately:

```python
try:
    risky_code()

except SomeError:
    handle_error()
```

```python
try:
    risky_code()

except SomeError:
    handle_error()

else:
    success_code()

finally:
    cleanup()
```
