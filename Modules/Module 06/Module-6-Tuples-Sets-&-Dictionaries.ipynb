# 🧩 Tuple

## Definition

A **tuple** is an **ordered** collection of items in Python that stores multiple values in a single variable. Unlike a list, a tuple is **immutable**, meaning its elements **cannot be changed, added, or removed** after the tuple is created. Tuples can hold **different data types**, including numbers, strings, booleans, and even other tuples. They are commonly used to represent **fixed data** that should remain unchanged throughout a program.

---

## Key Characteristics

* **Ordered:** ✅ Items maintain the order in which they were added.
* **Immutable:** 🔒 Elements cannot be modified, added, or removed after creation.
* **Indexed:** ✅ Each element has an index starting from `0`.
* **Allows Duplicate Values:** ✅ Duplicate elements are allowed.
* **Can Store Multiple Data Types:** Supports integers, floats, strings, booleans, objects, and nested tuples.
* **Syntax:** Created using **parentheses `()`** or simply by separating values with commas.
* **Lookup Time Complexity:** Approximately **O(1)** for accessing an element by its index.

---

## Syntax

```python
# Empty tuple
empty_tuple = ()

# Tuple with values
fruits = ("Apple", "Banana", "Mango")

# Tuple without parentheses (tuple packing)
numbers = 10, 20, 30

# Single-element tuple (comma is required)
single_item = (100,)

# Nested tuple
coordinates = ((10, 20), (30, 40))
```

---

## Example

```python
# Store student information in a tuple
student = ("Alice", 20, "Computer Science", 8.9)

# Access elements
print(student[0])    # Name
print(student[2])    # Course

# Length of the tuple
print(len(student))
```

**Output**

```python
Alice
Computer Science
4
```

## Real-Life Analogy

* Once the passport is issued, the information **cannot be changed** by simply editing it.
* You can **read** the information whenever needed, but you cannot modify it.

A Python **tuple** behaves in the same way—it stores data that should remain **constant and reliable**.

---

## When to Use

Use a **tuple** when:

* Your data should **never change** after creation.
* You want to represent **fixed records**, such as coordinates `(x, y)` or RGB color values `(255, 0, 0)`.
* You need to return **multiple values** from a function.
* You want to store **configuration settings** or constants.
* You need a collection that is generally **more memory-efficient** than a list for fixed data.

# 🎯 Set

## Definition

A **set** is an **unordered** collection of **unique** elements in Python. Unlike lists and tuples, a set **does not allow duplicate values**, making it ideal for storing distinct items. Sets are **mutable**, so you can add or remove elements after they are created. They are optimized for **fast membership testing** and support powerful mathematical operations such as **union**, **intersection**, **difference**, and **symmetric difference**.

---

## Key Characteristics

* **Ordered:** ❌ No. Elements do not have a fixed order.
* **Mutable:** ✅ Yes. Elements can be added or removed.
* **Indexed:** ❌ No. Elements cannot be accessed using an index.
* **Allows Duplicate Values:** ❌ No. Duplicate values are automatically removed.
* **Can Store Multiple Data Types:** ✅ Yes, as long as the elements are **hashable** (e.g., integers, strings, tuples).
* **Syntax:** Created using **curly braces `{}`** or the **`set()`** constructor.
* **Lookup Time Complexity:** Approximately **O(1)** for checking whether an element exists.

---

## Syntax

```python
# Empty set
empty_set = set()

# Set with values
fruits = {"Apple", "Banana", "Mango"}

# Create a set from a list
numbers = set([1, 2, 3, 4, 5])

# Set with different data types
mixed_set = {1, "Python", 3.14, (10, 20)}
```

> **Note:** `{}` creates an **empty dictionary**, **not** an empty set. Use `set()` to create an empty set.

---

## Example

```python
# A list containing duplicate values
numbers = [10, 20, 30, 20, 10, 40]

# Convert the list into a set
unique_numbers = set(numbers)

print(unique_numbers)

# Add a new element
unique_numbers.add(50)

# Remove an element
unique_numbers.remove(20)

# Check if a value exists
print(30 in unique_numbers)

print(unique_numbers)
```

**Output**

```python

```

> **Note:** The order of elements in the output may vary because **sets are unordered**.

# 📖 Dictionary

## Definition

A **dictionary** is a **mutable** data structure in Python that stores data as **key–value pairs**. Each **key** is unique and is used to access its corresponding **value**, making data retrieval fast and efficient. Dictionaries can store different data types, including numbers, strings, lists, tuples, and even other dictionaries. They are one of the most powerful and commonly used data structures for organizing and managing structured data.

---

## Key Characteristics

* **Ordered:** ✅ Yes. Dictionaries preserve the insertion order of key–value pairs (Python 3.7+).
* **Mutable:** ✅ Yes. Keys and values can be added, updated, or removed.
* **Indexed:** ❌ No. Values are accessed using **keys**, not numeric indexes.
* **Allows Duplicate Keys:** ❌ No. Every key must be unique.
* **Allows Duplicate Values:** ✅ Yes. Multiple keys can have the same value.
* **Can Store Multiple Data Types:** ✅ Keys must be **immutable** (e.g., strings, integers, tuples), while values can be of any data type.
* **Syntax:** Created using **curly braces `{}`** with **`key: value`** pairs.
* **Lookup Time Complexity:** Approximately **O(1)** for accessing a value using its key.

---

## Syntax

```python
# Empty dictionary
empty_dict = {}

# Dictionary with key-value pairs
student = {
    "name": "Alice",
    "age": 20,
    "course": "Computer Science"
}

# Dictionary with different value types
employee = {
    "id": 101,
    "name": "John",
    "salary": 55000,
    "skills": ["Python", "SQL"],
    "active": True
}
```

---

## Example

```python
# Create a dictionary
student = {
    "name": "Alice",
    "age": 20,
    "course": "Computer Science"
}

# Access a value using its key
print(student["name"])

# Add a new key-value pair
student["city"] = "New York"

# Update an existing value
student["age"] = 21

# Remove a key-value pair
del student["course"]

print(student)
```

**Output**

```python
Alice
{'name': 'Alice', 'age': 21, 'city': 'New York'}
```

---

## Real-Life Analogy

Imagine a **physical dictionary**.

* You search for a **word** (the **key**).
* The dictionary returns its **meaning** (the **value**).
* Every word appears **only once**, but different words can have similar meanings.
* Instead of reading every page, you directly search using the word.

A Python **dictionary** works in the same way—you use a **key** to instantly retrieve its associated **value**.

---

## When to Use

Use a **dictionary** when:

* You need to store data as **key–value pairs**.
* You want **fast access** to values using meaningful keys.
* You are representing real-world objects, such as students, employees, products, or user profiles.
* You need to frequently add, update, or remove information.
* Your data has descriptive labels instead of numeric positions.

### 11. Is the copy shallow or deep

# Deep Copy vs Shallow Copy in Python

In Python, copying objects can be a little confusing, especially when we have lists inside lists.

This notebook explains:

- Assignment
- Shallow copy
- Deep copy

in a simple way.

---

## 1. Variables and Objects in Python

In Python, a variable is like a **name tag** attached to an object.

Example:

```python
a = [1, 2, 3]
```

Here, `a` is pointing to a list object:

```python
[1, 2, 3]
```

Now, if we write:

```python
b = a
```

Python does **not** create a new list.

It only gives the same list another name.

So both `a` and `b` point to the **same list**.

---

## 2. Example of Assignment

```python
a = [1, 2, 3]
b = a

b.append(4)

print(a)
print(b)
```

Output:

```python
[1, 2, 3, 4]
[1, 2, 3, 4]
```

### Why did `a` also change?

Because `a` and `b` are pointing to the same list.

Changing `b` also changes `a`.

---

## 3. What is Copying?

Copying means creating a new object similar to the old one.

There are two main types of copying in Python:

1. **Shallow Copy**
2. **Deep Copy**

The difference becomes important when we have nested objects, like a list inside another list.

Example:

```python
original = [[1, 2], [3, 4]]
```

This is a list containing two smaller lists.

Think of it like:

> A big box containing two small boxes.

---

# 4. Shallow Copy

A **shallow copy** creates a new outer object, but the inner objects are still shared.

We can create a shallow copy using:

```python
import copy

shallow = copy.copy(original)
```

---

## Example of Shallow Copy

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

print(original)
print(shallow)
```

Output:

```python
[[1, 2], [3, 4]]
[[1, 2], [3, 4]]
```

At first, both look the same.

But the important thing is what happens when we change something.

---

## 5. Changing the Outer List in Shallow Copy

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

shallow.append([5, 6])

print(original)
print(shallow)
```

Output:

```python
[[1, 2], [3, 4]]
[[1, 2], [3, 4], [5, 6]]
```

Here, only `shallow` changed.

### Why?

Because the outer list was copied.

So `original` and `shallow` are two different outer lists.

---

## 6. Changing the Inner List in Shallow Copy

```python
import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

shallow[0].append(100)

print(original)
print(shallow)
```

Output:

```python
[[1, 2, 100], [3, 4]]
[[1, 2, 100], [3, 4]]
```

Now both changed.

### Why?

Because shallow copy only copied the outer list.

The inner lists are still shared.

So:

```python
original[0]
shallow[0]
```

Both point to the same inner list.

---

## 7. Simple Analogy for Shallow Copy

Imagine you have a school bag.

Inside the bag there are:

- a pencil box
- a notebook
- a water bottle

Now you make a shallow copy of the bag.

You get a new bag.

But inside the new bag, Python does not create new items.

It only gives references to the same old items.

So if you write your name in the notebook, both bags will show the changed notebook because it is the same notebook.

That is shallow copy.

---

# 8. Deep Copy

A **deep copy** creates a completely separate copy.

It copies:

- the outer object
- all inner objects
- and inner objects of inner objects, if needed

We can create a deep copy using:

```python
import copy

deep = copy.deepcopy(original)
```

---

## Example of Deep Copy

```python
import copy

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

deep[0].append(100)

print(original)
print(deep)
```

Output:

```python
[[1, 2], [3, 4]]
[[1, 2, 100], [3, 4]]
```

Now only `deep` changed.

The original list stayed safe.

### Why?

Because deep copy created a completely new copy of everything.

The inner lists are also separate.

---

## 9. Simple Analogy for Deep Copy

Again imagine a school bag.

This time you make a deep copy.

Now you get:

- a new bag
- a new pencil box
- a new notebook
- a new water bottle

Everything is separate.

If you write in the new notebook, the old notebook does not change.

That is deep copy.

---

# 10. Comparison Table

| Type | What happens? | If inner object changes, does original change? |
|---|---|---|
| Assignment | Same object | Yes |
| Shallow Copy | New outer object, shared inner objects | Yes |
| Deep Copy | New outer object and new inner objects | No |

---

# 11. Simple One-Level List Example

If your list is simple, like this:

```python

```

Then shallow copy is usually enough.

```python

b.append(4)

print(a)
print(b)
```

Output:

```python
[1, 2, 3]
[1, 2, 3, 4]
```

Here shallow copy works like a normal copy because there are no inner lists.

You can also use:

```python

```

```python

```

These also create shallow copies.

# 12. Full Example Comparing Assignment, Shallow Copy, and Deep Copy

```python
import copy

original = [[1, 2], [3, 4]]

assigned = original
shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0].append(99)

print("Original:", original)
print("Assigned:", assigned)
print("Shallow:", shallow)
print("Deep:", deep)
```

Output:

```python
Original: [[1, 2, 99], [3, 4]]
Assigned: [[1, 2, 99], [3, 4]]
Shallow: [[1, 2, 99], [3, 4]]
Deep: [[1, 2], [3, 4]]
```

### Explanation

- `assigned` changes because it is the same object as `original`.
- `shallow` changes because its inner lists are shared with `original`.
- `deep` does not change because everything is copied separately.

# 14. Important Point About Immutable Objects

Strings and numbers are immutable.

That means they cannot be changed after creation.

Example:

```python
name = "Rahul"
```

You cannot directly change the internal characters of this string.

So shallow copy usually does not create problems with strings and numbers.

But lists and dictionaries are mutable.

They can be changed.

So shallow copy can be confusing when lists or dictionaries are inside other lists or dictionaries.

---

# 15. When to Use Shallow Copy

Use shallow copy when:

- Your list contains simple values like numbers or strings.
- You do not care about inner objects being shared.
- You want faster and lighter copying.

Example:

```python
numbers = [1, 2, 3]
new_numbers = numbers.copy()
```

---

# 16. When to Use Deep Copy

Use deep copy when:

- You have nested lists, nested dictionaries, or complex objects.
- You want a completely independent copy.
- You want to change inner parts without affecting the original.

Example:

```python
import copy

data = [[1, 2], [3, 4]]
new_data = copy.deepcopy(data)
```

---

# 17. Short Summary

If you write:

```python
b = a
```

Then `b` and `a` are the same.

If you write:

```python
b = copy.copy(a)
```

Then the outer object is copied, but inner objects are shared.

If you write:

```python
b = copy.deepcopy(a)
```

Then everything is copied completely.

# 18. One-Line Memory Trick

**Assignment:** Same object.

**Shallow copy:** New outer object, shared inner objects.

**Deep copy:** Fully separate copy.

# 19. Final Test Example

```python
import copy

a = [[1, 2], [3, 4]]

b = a
c = copy.copy(a)

[[1, 2], [3, 4]]
```

### Why?

- `a` changed.
- `b` changed because it is the same as `a`.
- `c` changed because inner lists are shared.
- `d` did not change because it is a full deep copy.

---

## Final Conclusion

> **Shallow copy copies the main container but not the things inside it.**

> **Deep copy copies everything, including the things inside it.**

````python
# Shallow copy exists For mutable data types only
l1 = [1,2,[23,5,65],'python']
# To copy the list
import copy
# shallow copy
l2 = copy.copy(l1) # different memory location
print(l2)
print(f"l1 memory addres -> ",id(l1))
print(f"l2 memory addres -> ",id(l2))

# when the original list is changed/updated the shallow copy remains unchanged
l1[0] = 4
print(f"l1-> {l1}",id(l1))
print(f"l2-> {l2}"),id(l2)

# But when the shallow copy list is changed/updated the original list is also changed/updated 
l2[2][1] = 43
print(f"original list(l1) ->     {l1}",id(l1))
print(f"shallow copy list(l2) -> {l2}", id(l2))
````
