# 1. Installing Python

Before writing Python programs, we need to install Python on our computer.

Python provides the interpreter that reads and executes Python code.

---

## What is Python?

Python is a **high-level, general-purpose programming language**.

It was created by **Guido van Rossum** and was first released in 1991.

Python is designed to be:

* Easy to read
* Easy to write
* Beginner-friendly
* Powerful
* Flexible
* Cross-platform

A simple Python program looks like this:

```python
print("Hello, Python!")
```

Output:

```python
Hello, Python!
```

---

## Why Do We Need to Install Python?

When we write:

```python
print("Hello")
```

the computer does not automatically understand Python code.

The Python interpreter is required to:

1. Read the Python code.
2. Understand the instructions.
3. Execute the instructions.
4. Produce the output.

So, installing Python gives us the tools required to run Python programs.

---

## Checking Whether Python Is Already Installed

Open your terminal or command prompt.

On Windows, you can open:

```text
Command Prompt
```

or:

```text
PowerShell
```

Then type:

```bash
python --version
```

You may see something like:

```text
Python 3.12.5
```

You can also try:

```bash
python -V
```

Both commands display the installed Python version.

---

## Checking Python on Some Systems

On some Linux and macOS systems, you may need:

```bash
python3 --version
```

Example:

```text
Python 3.12.5
```

### Why might `python` and `python3` be different?

Some operating systems may use:

```bash
python3
```

to specifically refer to Python 3.

# 2. Installing Python on Windows

The general installation process is:

1. Download Python.
2. Start the installer.
3. Enable Python in PATH.
4. Install Python.
5. Verify the installation.

---

## Step 1: Download Python

Visit the official Python website:

[Python.org](https://www.python.org/?utm_source=chatgpt.com)

Download a current Python 3 release suitable for your operating system.

---

## Step 2: Start the Installer

After downloading Python, open the installer.

You will see an installation window.

One important option is:

```text
Add python.exe to PATH
```

Make sure this option is enabled before continuing.

---

## Why Is PATH Important?

PATH allows the operating system to find Python from the terminal.

Without PATH configuration, typing:

```bash
python
```

may not work from every terminal location.

So, enabling:

```text
Add python.exe to PATH
```

makes working with Python easier.

---

## Step 3: Install Python

Click the installation option and allow the installer to complete.

After installation, open a new terminal.

Run:

```bash
python --version
```

If Python was installed correctly, you should see its version.

---

## Step 4: Test Python

Run:

```bash
python
```

You may see something similar to:

```text
Python 3.x.x
>>>
```

The:

```text
>>>
```

is called the **Python prompt**.

You can now enter Python code directly.

For example:

```python
print("Hello, Python!")
```

Output:

```text
Hello, Python!
```

To exit the Python interpreter, you can use:

```python
exit()
```

# 29. Common Beginner Mistakes

## Mistake 1: Python Is Not Installed

Running:

```bash
python --version
```

may produce an error if Python is not installed or is not correctly configured.

---

## Mistake 2: Python Is Not Added to PATH

On Windows, the terminal may not recognize:

```bash
python
```

if Python was not added to PATH.

---

## Mistake 3: Confusing Python with PyCharm

Remember:

```text
Python → Programming language + interpreter
PyCharm → IDE
```

---

## Mistake 4: Wrong File Extension

Python files normally use:

```text
.py
```

For example:

```text
program.py
```

not:

```text
program.txt
```

---

## Mistake 5: Forgetting Indentation

Incorrect:

```python
if age >= 18:
print("Adult")
```

Correct:

```python
if age >= 18:
    print("Adult")
```

# 34. Final Test Example
