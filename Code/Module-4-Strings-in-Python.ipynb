 **Print function :**
- it is a built-in function that prints a message ,value or any other data type to the console or standard output. It can take multiple arguments and can be customized with various parameters such as 'sep' for separator, 'end' for ending character, and 'file' for specifying the output stream.



```python
# Print function : it is a built-in function that prints a message ,value or any other data type to the console or standard output. It can take multiple arguments and can be customized with various parameters such as 'sep' for separator, 'end' for ending character, and 'file' for specifying the output stream.

print("Hello, World!")  # Prints a simple message

print("The sum of 2 and 3 is:", 2 + 3)  # Prints a message along with the result of an expression

print("Python", "is", "fun", sep="-")  # Prints with a custom separator

print("This is the first line.", end=" ")  # Prints without a newline at the end

print("This is the second line.")  # Continues on the same line due to the previous print statement's 'end' parameter

print("Output to a file:", file=open('output.txt', 'w'))  # Prints to a specified file instead of the console
print("This will be written to the file 'output.txt'.")  # This line will not be printed to the console
print("Check the 'output.txt' file for the output.")  # Informs the user to check the file for output
print("This is the last print statement.")  # Final print statement to indicate the end of the script
```

**length of string**
- The length of a string is the number of characters it contains, including letters, numbers, spaces, and special characters. In Python, you can use the built-in len() function to determine the length of a string.

```python

s1 = "hello world"
print(s1)
print("The length of string is : " ,len(s1))
```

**Indexing in Strings**
- indexing is a way to access individual characters in a string using their position or index. In Python, strings are zero-indexed, which means that the first character of the string is at index 0, the second character is at index 1, and so on. You can also use negative indexing to access characters from the end of the string, where -1 refers to the last character, -2 refers to the second-to-last character, and so on.
- The string index always starts with 0, so the first character of the string is at index 0, the second character is at index 1, and so on.
- That means the last character of the string is at index len(string)-1.

```python

print("The character at index 0 is : " ,s1[0]) # so we can easily tell that the first character of the string is 'h' and it is at index 0.
print("The character at index 6 is : " ,s1[6]) # so we can easily tell that the character at index 6 is 'w' and it is at index 6.
print("The character at index -1 is : " ,s1[-1]) # so we can easily tell that the last character of the string is 'd' and it is at index -1.
print("The character at index 5 is : " ,s1[5]) # so we can easily tell that the space is also counted as a character in the string and it is at index 5.
```

### Slicing in Python

- **Slicing in string**
- Slicing is a way to extract a portion of a string by specifying a range of indices
- slicing syntax is string[start:end], where start is the index of the first character you want to include in the slice, and end is the index of the first character you want to exclude from the slice. The resulting slice will include all characters from start up to, but not including, end.
- start : is the first index of the slice. If start is omitted, it defaults to 0.
- end : is the last index of the slice. If end is omitted, it defaults to the length of the string.
- step : is the step size of the slice. If step is omitted, it defaults to 1.

```python

s1 = "Hello pyhon"
print("The substring from index 0 to 5 is : " ,s1[0:5])
print("The substring from index 6 to 11 is : " ,s1[6:11])
print("The substring from index -6 to -1 is : " ,s1[-6:-1]) # here -6 is the index of a whitespace.
print("The substring from index -5 to -1 is : " ,s1[-5:-1]) # here -5 is the index of the character 'p'.

print("The substring from index 0 to 11 with step size 2 is : " ,s1[0:11:2])
```

### fstring in python

- *fstring is a way to format strings in Python by embedding expressions inside string literals. It allows you to include variables or expressions directly within the string, making it easier to create dynamic and readable output. F-strings are denoted by prefixing the string with the letter 'f' or 'F' and using curly braces {} to enclose the expressions you want to evaluate.*

```python

name = "Kamraan"
age = 25
language = "Python"
nation = "USA"
color = "green"

print(name , "is", age, "years old.")

print(f"My name is {name} i am {age} years old.I am from {nation} and i like the {color} color . I am learning {language}.")

sub1 = 85
sub2 = 90
sub3 = 78
print(f"My name is {name} and my marks are {(sub1 + sub2 + sub3)} in three subjects and my average marks are {(sub1 + sub2 + sub3)/3}.")
```

### 3. Escape squance in python 

- Escape sequences are special characters that are used to represent certain characters or actions within a string. They are typically preceded by a backslash (\) and have a specific meaning. Here are some commonly used escape sequences in Python
- \ n : Represents a newline character, which moves the cursor to the next line.
- \ t : Represents a tab character.
- \ \ : Represents a literal backslash character.
- \ ' : Represents a single quote character.
- \ " : Represents a double quote character.

[TuteDude](https://upskill.tutedude.com/course/lecture-python)

```python
# \n is an escape sequence in Python that represents a newline character. When used in a string, it causes the text following it to be displayed on a new line. It is commonly used to format output and create multi-line strings.
print("hello everyne. \nHow are you")

# \t is an escape sequence in Python that represents a tab character. When used in a string, it inserts a horizontal tab space, which is typically equivalent to a few spaces (often 4 or 8, depending on the environment). It is commonly used to format output and create indented text.
print("hello everyne. \tHow are you")

# \\ is an escape sequence in Python that represents a single backslash character. Since the backslash is used to introduce escape sequences, you need to use a double backslash (\\) to include an actual backslash in a string. It is commonly used when you want to display or work with file paths or other strings that require a literal backslash.
print("This is a backslash: \\")
print("C:\\Users\\Username\\Documents")  # Example of a file path with backslashes

# \' is an escape sequence in Python that represents a single quote character. It is used to include a single quote within a string that is enclosed by single quotes, allowing you to avoid syntax errors. For example, if you want to include an apostrophe in a string, you can use the escape sequence \'.
print('It\'s a beautiful day!')  # Example of using \' to include an apostrophe in a string enclosed by single quotes

# print('This is a python's class')
print("This is a python's class")  # Example of using double quotes to include an apostrophe in a string


# print("He says, "Python is awesome!" ")  

print("He says, \"Python is awesome!\" ")  # Example of using \" to include double quotes in a string enclosed by double quotes
```

### 4. Operations on strings - membership, strip, replace

**Operations on strings**
- **membership operators** in Python are used to test whether a value or variable is found in a sequence (such as a string, list, tuple, or set). The two membership operators are 'in' and 'not in'. The 'in' operator returns True if the specified value is present in the sequence, while the 'not in' operator returns True if the specified value is not present in the sequence.

- **strip()** is a built-in string method in Python that removes leading and trailing whitespace characters (spaces, tabs, newlines) from a string. It returns a new string with the whitespace removed, while leaving the original string unchanged. If you want to remove specific characters instead of whitespace, you can pass those characters as an argument to the strip() method.

- **replace()** is a built-in string method in Python that allows you to replace occurrences of a specified substring with another substring within a string. It returns a new string with the replacements made, while leaving the original string unchanged. The method takes two arguments: the substring to be replaced and the substring to replace it with. You can also specify an optional third argument to limit the number of replacements.

```python
s = "Hello python"
print("The substring from index 0 to 5 is : " ,s[0:5])
print(s[0])
print(s[-1])
print(s)

language = "Python "
version = "3.9"
print(language + version)
```

```python
S = "Python "
print(S * 3)  # This will print "Python" three times in a row
# in strings '*' repetition operator is used to repeat the string a specified number of times. In this case, S * 3 means that the string "Python" will be repeated three times, resulting in the output "PythonPythonPython".
```

```python
# membership operator
# in
S = "Python is fun"
print("P" in S)  # This will print True
print("p" in S)  # This will print False
print("Py" in S)  # This will print True
print("thon" in S)  # This will print True
print("Java" in S)  # This will print False

#  not in
print("Python" not in S)  # This will print False
print("Java" not in S)  # This will print True
```

```python
# comparison of strings
str1 = "apple"
str2 = "banana"
print(str1 == str2)  # This will print False

print("apple" < "banana")  # This will print True because "apple" comes before "banana" lexicographically
print("banana" > "apple")  # This will print True because "banana" comes after "apple" lexicographically

print("apple " == "apple")  # This will print False because there is an extra space in the first string
```

```python
# Removing space from the string using strip()
S = "  Python  "
print(S.strip())  # This will print "Python" with leading and trailing spaces removed

s1 = " Python"
s2 = "Python"
print(s1==s2)  # This will print False because there is a leading space in s1
print(s1.strip() == s2)  # This will print True
```

```python
# replace()
s1 = "Hello world"
s2 = s1.replace("world", "Python")
print(s2)  # This will print "Hello Python"
# original string remains unchanged so that is way we can say that strings are immutable in python.
# we need to assign the result of the replace() method to a new variable or back to the original variable if we want to keep the changes.

s3 = s1.replace("o", "a")  # This will replace all occurrences of 'e' with 'a' in the string
print(s3)  # This will print "Halloca world"
s4 = s1.replace("o", "a", 1)  # This will replace only the first occurrence of 'o' with 'a' in the string
print(s4)  # This will print "Hella world"
```

### 5. Operation on strings -count, case, start, end

```python
# counting subtrings from a string using count()
s1 = "Hello world Hello Python"
s2 = "world"
s3 = " "
print(s1.count("Hello"))  # This will print 2
print(s1.count("o"))  # This will print 4
print(s1.count(s2))  # This will print 1
print(f"Occurrences of '{s2}' in the string is : {s1.count(s2)}")  # This will print "Occurrences of 'world' in the string: 1"

print(s1.count(s3))  # This will print 3, as there are three spaces in the string s1
```

```python
# change case of string using upper() , lower() ,title() and capitalize()
s1 = "hello worlD"
print(s1.upper())  # This will print "HELLO WORLD" i,e all characters are converted to uppercase.
print(s1.lower())  # This will print "hello world" i,e all characters are converted to lowercase.
print(s1.title())  # This will print "Hello World" i,e the first character of each word is capitalized and the rest are in lowercase.
print(s1.capitalize())  # This will print "Hello world" i,e the first character of the string is capitalized and the rest are in lowercase.

lang = "python 3.9"
print(lang.upper())  # This will print "PYTHON 3.9"
print(lang.lower())  # This will print "python 3.9"
```

```python
# sratswith() and endswith() methods are used to check if a string starts or ends with a specific substring. They return True if the condition is met, and False otherwise.
s1 = "Hello world we are learning python"
print(s1.startswith("Hello"))  # This will print True
print(s1.endswith("python"))  # This will print True
print(s1.startswith("world"))  # This will print False
print(s1.startswith("Hello world"))  # This will print True
print(s1.endswith(" are learning python"))  # This will print True
print(s1.startswith("Hello", 0, 5))  # This will print True, as it checks if the substring from index 0 to 5 starts with "Hello"
print(s1.endswith("python", 0, 30))  # This will print True, as it checks if the substring from index 0 to 30 ends with "python"
print(s1.startswith("world", 6, 11))  # This will print True, as it checks if the substring from index 6 to 11 starts with "world"
print(s1.endswith("on"))  # This will print True, as it checks if the string ends with "on"
```
