# 📋 List

## Definition

A **list** is an **ordered** collection of items in Python that can store multiple values in a single variable. Lists are **mutable**, meaning you can add, remove, or modify elements after they are created. They can contain **different data types**, including numbers, strings, objects, and even other lists. Lists are one of the most commonly used data structures because they are flexible and easy to work with.

---

## Key Characteristics

* **Ordered:** ✅ Items maintain the order in which they were added.
* **Mutable:** ✅ Elements can be added, removed, or modified.
* **Indexed:** ✅ Each element has an index starting from `0`.
* **Allows Duplicate Values:** ✅ Duplicate elements are allowed.
* **Can Store Multiple Data Types:** Supports integers, floats, strings, booleans, objects, and even nested lists.
* **Syntax:** Created using **square brackets `[]`**.
* **Lookup Time Complexity:** Approximately **O(1)** for accessing an element by its index.

---

## Syntax

```python
# Empty list
empty_list = []

# List with values
fruits = ["Apple", "Banana", "Mango"]

# List containing different data types
student = ["Alice", 20, 8.9, True]

# Nested list
matrix = [[1, 2], [3, 4]]
```

---

## Example

```python
# Create a list of fruits
fruits = ["Apple", "Banana", "Mango"]

# Access an element
print(fruits[0])

# Modify an element
fruits[1] = "Orange"

# Add a new element
fruits.append("Grapes")

# Remove an element
fruits.remove("Apple")

print(fruits)
```

**Output**

```python
Apple
['Orange', 'Mango', 'Grapes']
```

## When to Use

Use a **list** when:

* You need a collection whose contents **change over time**.
* The **order of elements matters**.
* You need to frequently **add, remove, or modify** elements.
* You want to store **multiple values** in a single variable.
* You need a versatile data structure for iteration, sorting, filtering, or processing data.

 **slicing in list** 
- slicing is a way to extract a portion of a list (or other sequence types like strings and tuples) by specifying a start index and an end index. The syntax for slicing is list[start:end], where 'start' is the index of the first element to include in the slice, and 'end' is the index of the first element to exclude from the slice.

````python
List = [1,3,2,5,4,7,8,9,6,10]  # This is a list of integers.
print(len(List))  # This will print the length of the list, which is 10.
print(List[3:8:1])  # This will print the elements from index 3 to 8 (exclusive) with a step of 1: [5, 4, 7, 8, 9]. The slice [3:8:1] means start at index 3, go up to (but not including) index 8, with a step of 1.
print(List[5:9:2])  # This will print every second element from the first eight elements of the list: [1, 2, 4, 8]. The slice [5:9:2] means start at index 5, go up to (but not including) index 9, with a step of 2.
````

### List operations - extend( )

````python
# extending a list with another list can be done using the `extend()` method.It allows you to add all the elements of an iterable (like another list, tuple, or set) to the end of the list, unlike the `append()` method, which adds its argument as a single element to the end of the list. The `extend()` method iterates over its argument and adds each element to the list, extending the list.
#  The syntax is as follows:
# list_name.extend(iterable)


fruits = ['apple', 'banana', 'cherry']  # This is a list of fruits.
print("original list of fruits: ", fruits)  # This will print the original list of fruits.
'fruits.append("kiwi","appricot","orange")' # This will raise a TypeError because the append() method takes exactly one argument. To add multiple elements, you should use the extend() method instead.

fruits.extend(["kiwi", "apricot"])  # This will add the elements "kiwi", "apricot", and "orange" to the end of the fruits list.

'fruits.extend("kiwi", "apricot")' # This will raise a TypeError because the extend() method takes exactly one iterable argument. To add multiple elements, you should pass them as a single iterable (like a list or tuple).

fruits.append(["orange","cherry"])  # This will add the list ["orange", "cherry"] as a single element to the end of the fruits list.
print("list of fruits after appending a list: ", fruits)  # This will print the list of fruits after appending a list: ['apple', 'banana', 'cherry', 'kiwi', 'apricot', ['orange', 'cherry']].

print("updated list of fruits: ", fruits)  # This will print the updated list of fruits: ['apple', 'banana', 'cherry', 'kiwi', 'apricot', 'orange'].

# Adding elements of another list to the fruits list using extend() method
vegetables = ['carrot', 'broccoli', 'spinach']
fruits.extend(vegetables)
print("list of fruits  after extending with vegetables: ", fruits)
````

### List operations - remove( )

- Removing an element from a list can be done using the `remove()` method.
- The syntax is : 
 *list_name.remove(element)*

````python
# removing an element from a list can be done using the `remove()` method. The syntax is as follows:
# list_name.remove(element)
fruits = ['apple', 'banana', 'cherry']  # This is a list of fruits.

fruits.remove('banana')  # This will remove the element 'banana' from the fruits list.

print("list of fruits after removing 'banana': ", fruits)  # This will print the list of fruits after removing 'banana': ['apple', 'cherry'].

# remove() method removes the element from the original list and does not return a new list. If the element is not found in the list, it raises a ValueError.

'print(fruits.remove("kiwi"))'  # This will raise a ValueError because 'kiwi' is not in the fruits list. The remove() method does not return a value, so it will print None if the element is found and removed successfully.

fruitz = ['apple', 'banana', 'cherry', 'banana']  # This is a list of fruits.
fruitz.remove('banana')  # This will remove the first occurrence of 'banana' from the fruitz list.

print("list of fruitz after removing first occurrence of 'banana': ", fruitz)  # This will print the list of fruitz after removing the first occurrence of 'banana': ['apple', 'cherry', 'banana'].
# that means if there are multiple occurrences of the element in the list, only the first one will be removed. If you want to remove all occurrences of an element, you can use a loop or a list comprehension.


````

### List operations - pop( )

````python
# pop() method removes and returns the last item in the list by default.
#  You can also specify an index to remove and return an item at that index.
#  The syntax is as follows:
# list_name.pop(index)
List = [1, 2, 3, 4, 5]  # This is a list of integers.
popped_item = List.pop()  # This will remove and return the last item in the list, which is 5.
print("popped item: ", popped_item)  # This will print the popped item: 5.
print("list after popping the last item: ", List)  # This will print the list after popping the last item: [1, 2, 3, 4].
# The pop() method modifies the original list and returns the removed item. If you specify an index, it will remove and return the item at that index instead of the last item. If the list is empty, calling pop() will raise an IndexError.
# pop() method is useful when you want to remove an item from a list and use it immediately, as it returns the removed item.
# pop() method takes the index of the element to be removed as an argument. If no index is specified, it removes and returns the last item in the list. If the specified index is out of range, it raises an IndexError.
List.pop(-1)  # This will remove and return the last item in the list, which is 4.
````

## 4. List Operation -reverse( ), sort( ), count( ),  membership( )

###  List Operation -reverse( )

````python
Days_of_week = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]  # This is a list of days of the week.
print("list of Days_of_week before reversing: ", Days_of_week)  # This will print the list of days of the week.

# reverse() method reverses the elements of the list in place, meaning it modifies the original list and does not return a new list. The syntax is as follows:
# list_name.reverse()
Days_of_week.reverse()
print("list of days of the week after reversing: ", Days_of_week)  # This will print the list of days of the week after reversing: ['Sun', 'Sat', 'Fri', 'Thu', 'Wed', 'Tue', 'Mon'].

# sort() method sorts the elements of the list in ascending order by default. The syntax is as follows:
# list_name.sort(reverse=False)
nums = [4,2,6,5,9,7]
print("list of nums before sorting: ", nums)  # This will print the list of nums before sorting: [4, 2, 6, 5, 9, 7].
nums.sort()
print("list of nums after sorting: ", nums)  # This will print the list of nums after sorting: [2, 4, 5, 6, 7, 9]. 

nums.sort(reverse=True)  # This will sort the list of nums in descending order.
print("list of nums after sorting in descending order: ", nums)  # This will print the list of nums after sorting in descending order: [9, 7, 6, 5, 4, 2].

# count() method counts the number of occurrences of an element in the list. The syntax is as follows:
# list_name.count(element)
nums = [1, 2, 3, 4, 5, 1, 2, 1]  # This is a list of integers.
count_of_1 = nums.count(1)  # This will count the number of occurrences of the integer 1 in the nums list.
print("count of 1 in the list of nums: ", count_of_1) 


````
