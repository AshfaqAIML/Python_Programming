```python
def add(a,b):
  return a+b

# Positional arguments - passing the arguments in order of their positions
result = add(5, 3)
print("Result of addition:", result)
```

```python
# Default Arguments 
def add(a,b=3):
  return a+b
result = add(5)
print("Result of addition with default argument:", result)

result1 = add(5, 10)
print("Result of addition with both arguments provided:", result1)
```

```python
def add(a,b=4,c):
  print(f"a: {a}, b: {b}, c: {c}")
  return a+b+c

result = add(2,7,4)
print(result)
# non default arguments cannot follow default arguments. In the function definition, all non-default arguments must be defined before any default arguments. This is because when the function is called, Python needs to know how to assign values to the parameters, and having non-default arguments after default arguments creates ambiguity in the function call.
```

Variable Length argument *args - variabe length positional arguments

```python
def add(a,b=5,c=10):
  print(f"a: {a}, b: {b}, c: {c}")
  return a+b+c  
result = add(2,5,8,9)
print(result)

#----> 4 result = add(2,5,8,9)
#TypeError: add() takes from 1 to 3 positional arguments but 4 were given
```

```python
# so if we want to pass more arguments than the function definition allows, we can use *args to allow for a variable number of positional arguments. This way, we can handle any number of arguments without causing an error.
```

```python
# *args allows us to pass a variable number of positional arguments to a function. It collects all the additional positional arguments into a tuple, which can be accessed within the function. This is useful when we want to create functions that can handle an arbitrary number of arguments without having to define them explicitly in the function signature.

def add(*args):
  print("Arguments passed:", args)
  print("Type of args:", type(args))
  # to all numbers of args
  return sum(args)  
result = add(1, 2, 3)
print("Result of addition with *args:", result) 


add(1, 2, 3)
add()
```

```python
# we can use *nums instead of *args, the name after the asterisk is just a convention and can be any valid variable name. The important part is the asterisk (*) which indicates that the function can accept a variable number of positional arguments. The name following the asterisk is used to refer to the collected arguments within the function.
def add(*nums):
  print("Arguments passed:", nums)
  print("Type of nums:", type(nums))
  return sum(nums)
result = add(1, 2, 3, 4, 5)
print("Result of addition with *nums:", result)
```

```python
# Real-world example of *args
def students_details(stdid,stdnm,*marks):
  if len(marks) == 0:
    print(f"Student ID: {stdid}")
    print(f"Student Name: {stdnm}")
    print("Marks: No marks provided")
    print("Average Marks: N/A")
    print("student may be absent or marks not entered")
  else:
    print(f"Student ID: {stdid}")
    print(f"Student Name: {stdnm}")
    print("Marks:", marks)
    print("Average Marks:", sum(marks)/len(marks))
    # print("Average Marks:", sum(marks)/len(marks) if marks else 0)
students_details(101, "Alice", 85, 90, 78)
students_details(102, "Bob", 80, 85)
students_details(103, "Charlie",78,98,90,87,67,97,56,78,98)
students_details(104, "David")
```

6. Variable length keyword arguments
**kwargs

```python
# **kwargs allows us to pass a variable number of keyword arguments to a function. It collects all the additional keyword arguments into a dictionary, which can be accessed within the function. This is useful when we want to create functions that can handle an arbitrary number of named arguments without having to define them explicitly in the function signature.
```

```python
# **kwargs 
def student_details(sid,sname,**marks):
  if marks  == 0:
    print(f"Student ID: {sid}")
    print(f"Student Name: {sname}")
    print("No marks provided because student may be absent or marks not entered")
  else:
    print(f"Student ID: {sid}")
    print(f"Student Name: {sname}")
    print("Keyword Arguments (marks):", marks)
    print("Percentage Marks:", sum(marks.values())/len(marks) if marks else 0)

student_details(101, "Alice", math=85, science=90, english=78)
student_details(102, "Bob", math=80, science=85)


```

```python
def student_details(**marks,sid,sname):
  print(f"Student ID: {sid}")
  print(f"Student Name: {sname}")
  print("Keyword Arguments (marks):", marks)
  print("Percentage Marks:", sum(marks.values())/len(marks) if marks else 0)

student_details(sid=101, sname="Alice", math=85, science=90, english=78)

#  arguments cannot follow var-keyword argument in the function definition, all non-default arguments must be defined before any default arguments. This is because when the function is called, Python needs to know how to assign values to the parameters, and having non-default arguments after default arguments creates ambiguity in the function call.
```

```python
# def student_details(sid,sname,**marks,*args): not allowed bcz all the non-default arguments (sid and sname) must be defined before the var-keyword argument (**marks) and the var-positional argument (*args). This is because when the function is called, Python needs to know how to assign values to the parameters, and having non-default arguments after var-keyword or var-positional arguments creates ambiguity in the function call.
def student_details(sid,sname,*args,**marks):
  print(f"Student ID: {sid}")
  print(f"Student Name: {sname}")
  print("Keyword Arguments (marks):", marks)
  print("Percentage Marks:", sum(marks.values())/len(marks) if marks else 0)
```

```python
def student_details(sid,sname,*extra,**marks):
  if marks  == 0:
    print(f"Student ID: {sid}")
    print(f"Student Name: {sname}")
    print("No marks provided because student may be absent or marks not entered")
    print("Extra Carricular Activities:", extra)

  elif extra == 0:
    print(f"Student ID: {sid}")
    print(f"Student Name: {sname}")
    print("Keyword Arguments (marks):", marks)
    print("Percentage Marks:", sum(marks.values())/len(marks) if marks else 0)
    print("No extra carricular activities provided")
  else:
    print(f"Student ID: {sid}")
    print(f"Student Name: {sname}")
    print("Keyword Arguments (marks):", marks)
    print("Percentage Marks:", sum(marks.values())/len(marks) if marks else 0)
    print("Extra Carricular Activities:", extra)

student_details(sid=101, sname="Alice", extra=("Football", "Basketball"), math=85, science=90, english=78)
student_details(sid=102, sname="Bob", math=80, science=85)
student_details(sid=103, sname="Charlie", extra=("Chess", "Debate Club")) 

#TypeError: unsupported operand type(s) for +: 'int' and 'tuple' 
#---> 21 student_details(sid=101, sname="Alice", extra=("Football", "Basketball"), math=85, science=90, english=78)
#---> 18 print("Percentage Marks:", sum(marks.values())/len(marks) if marks else 0)
```
